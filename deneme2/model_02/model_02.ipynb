### TEK DUGME

```
dusun_gecis = 2     govde IKI kez kosar (model_00: 1)
dusun_tau   = 1.0   ara dagilimin sicakligi
dusun_alfa  = 1.0   geri gomulen z'nin agirligi
```

Gecisler arasinda, birinci gecisin gizli durumu jeton dagilimina
cevrilip **geri gomulur**:

```
p = softmax(E h / tau)      jetonlar uzerinde dagilim
z = E^T p                   YUMUSAK JETON -- gomme uzayinda
h = govde(x, ek = alfa*z)   ikinci gecis
```

**Kayip YALNIZ son gecise uygulanir.** Ara gecis DENETLENMEZ: ne
uretecegi serbest, tek kisit gomme satirlarinin DISBUKEY BIRLESIMI
olmak zorunda olmasi.

### NEDEN -- olculenlerden turedi

```
1. hop tek basina      1.0000      bilgi VAR
2. hop tek basina      0.9997      bilgi VAR
IKISI BIRDEN           0.0550
kopru dogrusal sondada 0.0482      taban 0.0361  -> e2 HIC yazilmiyor
h(e1,r1) ~ h(e2)      +0.0053      ortanca sira 132/500
iliski dogrusal operator MU        17'nin 11'inde HAYIR (R^2 ~ 0)
```

Son satir kritik: model 1-hop'u `1.0000` ile cevapliyor ama o
iliskilerin cogunda varliktan hedefe **dogrusal yol yok**. Yani olgular
bir donusum olarak degil **TABLO** olarak saklaniyor. Tablo, tanimi
geregi gorulmemis satiri uretemez.

```
model YAPABILIYOR   L(e, r) -> a        tek tablo aramasi
model YAPAMIYOR     L(L(e,r1), r2)      birincinin CIKTISINI ikincinin GIRDISI yapmak
```

Eksik olan devre degil, **TESISAT**. Bir tablo aramasi cevabi CIKIS
uzayinda uretir; ikinci arama anahtarini GIRIS uzayinda ister. CoT bu
cevirmeni jetonu YAZIP GERI OKUYARAK yapar -- ve bu proje ortuk
cikarimi arastirdigi icin o yol kapali. Bagli gomme
(`head.weight is emb.weight`) ayni cevirmeni **jeton yazmadan** veriyor.

### NE DENENDI, NE OLDU

```
2-hop'u egitimden CIKARMAK   comp 0.2760 -> 0.0113   kisayol ALTI KAT
agirlik cezasi 5 KATINA      kisayol 0.4573 -> 0.5313, gerisi dusuyor
kopruyu ZORLAMAK (b8)        comp 0.9747, kisayol 0.0003  -> COZULEBILIR
```

Kisayolu **dolayli** yollarla kapatmak ise yaramiyor: kisayol bir kusur
degil, modelin sahip oldugu tek islem. Yerine koyacak bir sey vermeden
kapatmak geri kalani bozuyor. Bu kol, yerine konacak seyi veriyor.

````python
# 6 RAPOR (canli durum)  |  CPU  |  tekrar: GUVENLI
import json, glob, os, statistics

# KIYAS TABANI: model_00. AYNI MIMARI, AYNI OPTIMIZASYON, AYNI VERI,
# AYNI SINAV (iz d6751004648c), AYNI EGITIM HAVUZU. TEK FARK ILERI
# GECIS: `dusun_gecis` 1 -> 2. Dizi uzunlugu bile degismedi, o yuzden
# bu projedeki EN TEMIZ kiyas.
M00 = dict(one=1.0000, seen=1.0000, comp=0.2760, ood=0.0381,
           ent=0.0407, ent_yok=0.0450)      # model_00 PENCERE 72-80k
B15 = dict(one=0.9997, seen=0.9997, comp=0.0550, ood=0.0143,
           ent=0.0290, ent_yok=0.0409)      # model_b15 PENCERE 12-20k
B8 = dict(one=0.9510, seen=0.9913, comp=0.9747, ood=0.0177,
          ent=0.8650, ent_yok=0.8710)       # TAVAN: kopru EGITIMDE zorlandi
M00_KSY = 0.0760                            # model_00 kisayol -- BU KOLUN HEDEFI
M00_MS = 74.2                               # model_00 OLCULDU (t_len 17)

_TUM = (("adim", "adim"), ("kayip", "kayip"), ("one", "one"),
        ("seen", "seen"), ("comp", "comp"), ("ood", "ood"), ("ent", "ent"),
        ("ent_kati", "ent_kati"), ("ent_yok", "ent_yok"),
        ("kayip_ana", "kayip_ana"),
        ("ent_kisayol", "ent_ksy"), ("ent_yok_kisayol", "yok_ksy"))

for kl in sorted(k for k in glob.glob(f"{EV}/t*") if os.path.isdir(k)):
    eg = glob.glob(f"{kl}/egri_*.json")
    if not eg:
        print(f"{os.path.basename(kl)}: egri YOK"); continue
    ky = glob.glob(f"{kl}/kosu_t*.json")
    k = json.load(open(ky[0])) if ky else {}
    e = json.load(open(eg[0]))
    _var = set().union(*(set(r) for r in e))
    SUT = tuple(x for x in _TUM if x[0] in _var)
    _atlanan = [c for c in _var
                if c.startswith(("one", "seen", "comp", "ood", "ent"))
                and c not in {x[0] for x in SUT}]
    assert not _atlanan, f"OLCULUYOR AMA BASILMIYOR: {_atlanan}"
    _iz = k.get("olcme_izi", "?")
    assert _iz == "d6751004648c", (
        f"olcme izi {_iz} != d6751004648c -- model_00 KIYASI GECERSIZ")
    print("=" * 84)
    print(f"{os.path.basename(kl)}   {k.get('ad','?')}   durum {k.get('durum','?')}"
          f"   commit {k.get('commit','?')}   {k.get('gpu','')}")
    v = k.get("veri", {})
    if v:
        print(f"   olgu {v['olgu']}  egitim2 {v['egitim2']}  ENT {v['ent']}  "
              f"phi {v['phi']} (wang {v['wang_phi']})  "
              f"parametre {k.get('parametre',0):,}  iz {k.get('olcme_izi','?')}")
    print("   " + "".join(f"{b:>10}" for _, b in SUT) + f"{'dk':>6}")
    for r in e:
        hcr = []
        for c, _ in SUT:
            x = r.get(c)
            hcr.append(f"{x:>10d}" if c == "adim" else
                       (f"{'---':>10}" if x is None else f"{x:>10.4f}"))
        print("   " + "".join(hcr) + f"{r['sn']/60:>6.0f}")

    s = e[-1]
    print("   " + "-" * 81)
    print(f"   {'GECTI ' if s.get('one',0) >= 0.98 else '!! KALDI'}"
          f"  {'SAGLIK-1HOP':<14} one >= 0.98      <- ON KOSUL")
    print(f"   {'GECTI ' if s.get('seen',0) >= 0.95 else '!! KALDI'}"
          f"  {'SAGLIK-EZBER':<14} seen >= 0.95     <- bu kolda GECERLI")
    print(f"   {'GECTI ' if s.get('comp',0) >= 0.50 else '!! KALDI'}"
          f"  {'OLGUNLUK':<14} comp >= 0.50")
    print(f"   {'GECTI ' if abs(s.get('ent_yok_kisayol',1)) < 1e-9 else '!! KALDI'}"
          f"  {'BIRIM TESTI':<14} ent_yok_kisayol == 0")

    print("   " + "-" * 81)
    d = [(b["sn"] - a["sn"]) / (b["adim"] - a["adim"]) * 1000
         for a, b in zip(e, e[1:])
         if b["sn"] > a["sn"] and b["adim"] > a["adim"]]
    if d:
        ms = statistics.median(d)
        print(f"   HIZ ortanca {ms:.1f} ms/adim   (model_00 {M00_MS})"
              f"   {(ms/M00_MS-1)*100:+.1f}%   [govde IKI kez kosuyor]")
    print(f"   {'olcu':<9}{'b15 pen':>10}{'model_00':>11}{'model_02':>10}"
          f"{'b8 TAVAN':>10}")
    for c in ("one", "seen", "comp", "ood", "ent", "ent_yok"):
        if c in s:
            print(f"   {c:<9}{B15[c]:>10.4f}{M00[c]:>11.4f}{s[c]:>10.4f}"
                  f"{B8[c]:>10.4f}")
    print("   model_00 sutunu PENCERE okumasi (72-80k). model_02 ile")
    print("   model_00 arasinda TEK fark dusun_gecis -- ayni veri, ayni")
    print("   havuz, ayni dizi uzunlugu, AYNI SINAV.")
    if "comp" in s:
        c_ = s["comp"]
        print("   BIRINCIL (comp -- onkayit model_02.md): "
              + ("OLGUNLUK KAPISI GECTI" if c_ >= 0.50 else
                 "0.30-0.50 BELIRGIN kazanc -- tesisat hipotezi AYAKTA"
                 if c_ >= 0.30 else
                 "0.10-0.30 model_00 mertebesi -- dongu KAZANC VERMEDI"
                 if c_ >= 0.10 else
                 "<0.10 dongu ZARARLI"))
        print("      ON KOSUL: `one` kapisi. Gecmezse comp YORUMLANMAZ.")
        print(f"      model_00 comp {M00['comp']:.4f} -> model_02 {c_:.4f}"
              f"   fark {c_-M00['comp']:+.4f}")
    if "ent_kisayol" in s:
        ksy = s["ent_kisayol"]
        print(f"   KISAYOL ent_ksy {ksy:.4f}  (model_00 {M00_KSY:.4f})"
              f"   fark {ksy-M00_KSY:+.4f}")
        print("      Dongu kopruye yol actiysa kisayol DUSMELI. Yukseldiyse")
        print("      ara gecis kopruyu DEGIL cevabi tahmin ediyor olabilir --")
        print("      `p1` entropisi bunu ayirir (11. hucre, --dusun).")
    if "ent" in s:
        print(f"   ent {s['ent']:.4f} (model_00 {M00['ent']:.4f})"
              f"  -- OLGUNLUK gecmezse YORUMLANMAZ")
    if "ood" in s:
        print(f"   ood {s['ood']:.4f} -- RAPORLANIR, HUKUM VERMEZ. 210 ornek.")
````
