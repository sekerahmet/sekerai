### KARAR KURALI (önkayıttan)  ·  ⛔ *hedefler BAŞKA SINAVDAN, geçersiz*

Hedef — `model_07` pencere okuması:

```
t12-20k   comp 0.5017   ent 0.3447
t32-40k   comp 0.7436   ent 0.5447
t52-60k   comp 0.8135   ent 0.6547
```

`comp` ve `ent` için AYRI AYRI, her seviyeye ilk ulaşılan adım:

```
N(08) < N(07) - 4000     HIZLANDIRDI
|N(08) - N(07)| <= 4000  AYIRT EDILEMEDI
N(08) > N(07) + 4000     YAVASLATTI
20.000'de ulasilmadi     O SEVIYE ICIN HUKUM YOK
```

±4000 = ölçüm ızgarasının (2000) iki katı. Koşudan önce sabitlendi.
